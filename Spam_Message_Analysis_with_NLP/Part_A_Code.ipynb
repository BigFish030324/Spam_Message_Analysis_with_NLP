## Question 1: Tokenization and Text Filtering

Different libraries will be used to perform tokenization, as well as stop words and punctuation removal on `data_1`.

### Split Function

```python
# Retrieve the input text
input_text = data_1

# Create a list to store the tokenized words
split_tokens = []

# Document contains new lines, so loop through each line
for line in input_text.splitlines():

    # Split the line based on blank spaces
    split_line = line.split(" ")

    # Then only attach the split words to the list
    split_tokens.extend(split_line)

# Lastly, output the results
print(split_tokens)
```

### Regular Expressions

```python
# Import library for regular expression
import re

# Retrieve the input text
input_text = data_1

# Perform splitting based on words
# \b means word boundary
# \w+ means one or more word character (letters, numbers, underscore)
regex_tokens = re.findall(r'\b\w+\b', input_text)
print(regex_tokens)
```

### NLTK Library

```python
# Import library for NLTK
import nltk

# Retrieve the input text
input_text = data_1

# Perform splitting using NLTK's word_tokenize function
nltk_tokens = nltk.tokenize.word_tokenize(input_text)
print(nltk_tokens)
```

### Improvement for Tokenization for Split Function and Regular Expressions

```python
# Import libraries for regular expression
import re

# Retrieve the input text
input_text = data_1

# Split function
clean_text = re.sub(r'([.,;:!?()])', r' \1 ', input_text)               # Have to insert blank spaces around punctuation first
split_tokens = clean_text.split()
print("\nSplit function:")
print(split_tokens)

# Regex method
regex_tokens = re.findall(r"\w+(?:-\w+)*|[.,;:!?()]", input_text)       # Handle hyphens (e.g., "open-class") and punctuations
print("\nRegex method:")
print(regex_tokens)
```

### Stop Words and Punctuation Removal

```python
# Library import
from nltk.corpus import stopwords
import nltk
import string           # string library is used to handle punctuations

# Download the stop word corpus and extract the English stop words
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

# Retrieve the NLTK tokens from previous step
tokens = nltk_tokens

# Separate the tokens into clean tokens, stop words and punctuations
clean_tokens = []
found_stop_words = []
found_punctuations = []

# Loop through each token to categorize them
for token in tokens:

    token = token.lower()       # Convert to lowercase so that tokens can match the stop words from corpus

    if token in stop_words:
        found_stop_words.append(token)

    elif token in string.punctuation:
        found_punctuations.append(token)

    else:
        clean_tokens.append(token)

# Output the results
print("Clean Tokens:")
print(clean_tokens)

print("\nStop Words Found:")
print(set(found_stop_words))

print("\nPunctuations Found:")
print(set(found_punctuations))
```

## Question 2: Word Stemming

### Stemming using Regular Expression

```python
# A function to remove suffix using regex
def regex_stem(word):
    return re.sub(r"(ing|ed|ly|es|s|ion)$", "", word)

# Apply stemming using regex for each cleaned token (from Q1)
regex_stemmed_tokens = [regex_stem(token) for token in clean_tokens]

# Output and compare the results
print("Original Clean Tokens:")
print(clean_tokens)
print("\nRegex Stemmed Tokens:")
print(regex_stemmed_tokens)
```

### Porter Stemmer

```python
# Import library for PorterStemmer
from nltk.stem import PorterStemmer

# Create an instance of PorterStemmer
porter_stemmer = PorterStemmer()

# Perform stemming on each cleaned token using PorterStemmer
porter_stemmed_tokens = [porter_stemmer.stem(token) for token in clean_tokens]

# Output and compare the results
print("Original Clean Tokens:")
print(clean_tokens)
print("\nPorter Stemmed Tokens:")
print(porter_stemmed_tokens)
```

### Lancaster Stemmer

```python
# Import library for LancasterStemmer
from nltk.stem import LancasterStemmer

# Create an instance of LancasterStemmer
lancaster_stemmer = LancasterStemmer()

# Perform stemming on each cleaned token using LancasterStemmer
lancaster_stemmed_tokens = [lancaster_stemmer.stem(token) for token in clean_tokens]

# Output and compare the results
print("Original Clean Tokens:")
print(clean_tokens)
print("\nLancaster Stemmed Tokens:")
print(lancaster_stemmed_tokens)
```

### POS Tagging using Regular Expression

```python
# NLTK for tokenization
import nltk
from nltk.tokenize import word_tokenize

# Retrieve data
text_data = data_2

# First, perform tokenization to divide sentences into words
tokens = word_tokenize(text_data)

# Define the RE patterns to be used
patterns = [

    (r'^\d+(\.\d+)?$', 'CD'),                           # Cardinal number (integer or decimal)

    (r'.*ed$', 'VBD'),                                  # Past tense verbs (e.g., walked, barked)
    (r'.*ing$', 'VBG'),                                 # Gerunds or present participles (e.g., walking, barking)
    (r'.*es$', 'VBZ'),                                  # 3rd person singular present verbs (e.g., watches, goes)

    (r'.*ly$', 'RB'),                                   # Adverbs ending with -ly (e.g., quickly, silently)
    (r'.*ful$', 'JJ'),                                  # Adjectives ending with -ful (e.g., beautiful, helpful)
    (r'.*able$', 'JJ'),                                 # Adjectives ending with -able (e.g., readable, comfortable)
    (r'.*ous$', 'JJ'),                                  # Adjectives ending with -ous (e.g., dangerous, curious)

    (r'.*big$', 'JJ'),                                  # Specific adjective: big
    (r'.*black$', 'JJ'),                                # Specific adjective: black
    (r'.*white$', 'JJ'),                                # Specific adjective: white

    (r'^(the|a|an)$', 'DT'),                            # Determiners / articles (e.g., the, a, an)
    (r'^(and|or|but)$', 'CC'),                          # Coordinating conjunctions (e.g., and, or, but)
    (r'^(at|in|on|by|with|from|to|of|for)$', 'IN'),     # Prepositions (e.g., at, in, on, by, with, from, to, of, for)

    (r'\.$', '.'),                                      # Full stop punctuation
    (r'.*', 'NN')                                       # Default to noun (NN) for any word that doesn't match the above patterns
]

# Use RegexpTagger to tag each word with the defined pattern and output results
regex_tagger = nltk.RegexpTagger(patterns)
regex_tagging = regex_tagger.tag(tokens)
print(regex_tagging)
```

### Parse Tree using NLTK

```python
# Import NLTK library for prase tree generation
import nltk

# Define the CFG
cfg = nltk.CFG.fromstring("""
S -> NP VP
NP -> DT JJ JJ NN | DT JJ NN
VP -> VBD PP | VBD RB | VP CC VP
PP -> IN NP

DT -> 'The' | 'the'
JJ -> 'big' | 'black' | 'white'
NN -> 'dog' | 'cat'
VBD -> 'barked' | 'chased'
IN -> 'at'
RB -> 'away'
CC -> 'and'
""")

# Retrieve data and eliminate full stop
sentence = data_2.replace('.', '')

# Create the parser from NLTK
parser = nltk.ChartParser(cfg)
for tree in parser.parse(nltk.word_tokenize(sentence)):
   display(tree)
```

## Question 4: Calculating Sentence Probability using Bigram Models

```python
# Import libraries for bigram models
from nltk.lm.preprocessing import padded_everygram_pipeline
from nltk.lm import MLE, Laplace

# Initialize the corpus and the test sentence
corpus = [
    "He read a book",
    "I read a different book",
    "He read a book by Danielle",
]
corpus = [sentence.split() for sentence in corpus]
test_sentence = "I read a different book by Danielle".split()

# Set n as 2 for bigram models
n = 2

# Train unsmoothed model
mle_model = MLE(n)
mle_model.fit(*padded_everygram_pipeline(n, corpus))

# Train smoothed model
laplace_model = Laplace(n)
laplace_model.fit(*padded_everygram_pipeline(n, corpus))

# Calculate unsmoothed probability
prob_unsmoothed = 1
for index in range(len(test_sentence) + 1):
    prob_unsmoothed *= mle_model.score(
        "</s>" if index == len(test_sentence) else test_sentence[index],
        ["<s>"] if index == 0 else [test_sentence[index - 1]]
    )

# Calculate smoothed probability
prob_smoothed = 1
for index in range(len(test_sentence) + 1):
    prob_smoothed *= laplace_model.score(
        "</s>" if index == len(test_sentence) else test_sentence[index],
        ["<s>"] if index == 0 else [test_sentence[index - 1]]
    )

# Output results
print(f"Unsmoothed Sentence Probability: {prob_unsmoothed:.4f}")
print(f"Laplace-Smoothed Sentence Probability: {prob_smoothed:.10f}")
print(f"NLTK Vocabulary Size: {len(mle_model.vocab)}")
```
